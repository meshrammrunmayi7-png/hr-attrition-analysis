```python
# Import required libraries for data analysis
# Load HR attrition dataset into a DataFrame
# Preview first 5 rows to understand structure

import pandas as pd

df = pd.read_csv("IBM.csv")
df.head()
```

```python
# Review dataset structure, column names, row counts, and data types

df.info()
```

```python
# Check for missing values in each column

df.isnull().sum()
```

```python
# Replace spaces with underscores for easier coding and SQL queries
# Convert column names to lowercase for consistency

df.columns= df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
# Check for missing values in each column

df.duplicated().sum()
```

```python
# Generate summary statistics for numerical columns

df.describe()
```

```python
# Rename columns for consistency and easier readability

df= df.rename(columns=
    {
    "distancefromhome" : "distance_from_home",
    "environmentsatisfaction" : "environment_satisfaction",
    "jobsatisfaction" : "job_satisfaction",
    "maritalstatus" : "marital_status",
    "monthlyincome" : "monthly_income",
    "numcompaniesworked" : "num_companies_worked",
    "worklifebalance" : "work_life_balance",
    "yearsatcompany" : "years_at_company"
    }) 
    
```

```python
# View unique categories in attrition column

df["attrition"].unique()
```

```python
# Count employees who stayed vs left the company

df["attrition"].value_counts()
```

```python
# View unique department names

df["department"].unique()
```

```python
# Count number of employees in each department

df["department"].value_counts()
```

```python
# Compare attrition counts across departments

df.groupby("department")["attrition"].value_counts()
```

```python
# Calculate attrition percentage by department

pd.crosstab(df["department"],df["attrition"], normalize= "index").round(2) * 100
```

```python
# Export cleaned dataset for SQL and Tableau analysis

df.to_csv("cleaned_hr_data.csv", index=False)
```
